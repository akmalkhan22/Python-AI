```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(
    "ecommerce_sales_analytics_5000.csv",
    parse_dates = ["order_date"]
)
```

```python
df.head()
```

```python
prd_cat_sum = df.groupby("product_category")["revenue"].sum()

#display bar chart
fig, ax = plt.subplots()
ax.bar(prd_cat_sum.index, prd_cat_sum.values)

ax.set_xlabel("Category")
ax.set_ylabel("Total Revenue")
ax.set_title("Product sales by categories")
```

```python
group_product = df.groupby(["product_category", "region"])["revenue"].sum().unstack()
group_product.plot(kind = "bar", figsize = (12, 7))
plt.ylabel("Revenue")
plt.xlabel("Revenue by Categoryo and Region")
```

```python

group_product.plot(kind = "bar", stacked = True, figsize = (12, 7))
plt.ylabel("Revenue")
plt.xlabel("Revenue by Categoryo and Region")
```
