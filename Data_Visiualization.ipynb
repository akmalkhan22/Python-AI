```python
import matplotlib.pyplot as plt
```

```python
#demmy data
days = [1,2,3,4,5,6,7,8]
sales = [100, 400, 200, 300, 900,600, 700, 800]

#line graph
plt.plot(days, sales,
         marker = 'o',
         linewidth = 2,
         alpha = 0.8,
         label = "Daily sales")
plt.xlabel("days")
plt.ylabel("Sales")
plt.title("Daily Sales Graph")
plt.legend()
plt.show()
```

```python
#sample data
temperature = [20, 25, 30,35,25,40,30]
ice_cream_sale = [10,15,23,28,16,35,20]

plt.scatter(temperature, ice_cream_sale,
           marker = "x",
            alpha = 0.8,
            label = "sales"
            
           )

plt.xlabel("temperature")
plt.ylabel("ice_cream_sale")
plt.title("Ice cream sales")
plt.legend()
plt.show()
```

```python
import pandas as pd
```

```python
data = pd.read_csv("ecommerce_sales_analytics_5000.csv", parse_dates= ["order_date"])
df = data.sample(n=100)
```

```python
monthly_revenue =(
    df
    .set_index("order_date")['revenue']
    .resample("ME")
    .sum()
)

fig, ax = plt.subplots(figsize=(7,4))
ax.plot( monthly_revenue.index, monthly_revenue.values, linewidth =2)
ax.set_xlabel("Months")
ax.set_ylabel("Sales")
ax.set_title("Monthly sales")
```

```python
fig,ax = plt.subplots(figsize = (7,4))

ax.scatter(df['unit_price'], df['revenue'], alpha = 0.5)
ax.set_xlabel("Unit Price")
ax.set_ylabel("Revenue")
```
